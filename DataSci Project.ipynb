```python
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Summer"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    else:
        return "Autumn"

df["season"] = df["month"].apply(get_season)
```

## Limitations of the Study

1. **Limited city coverage:** The study covers only ten selected Indian cities, so the findings may not represent every city in India.

2. **Limited variables:** The analysis uses the weather variables available in the dataset. Factors such as land use, vegetation, building density, traffic, and satellite-based surface temperatures are not included.

3. **Historical data:** The study covers the period from 2000 to 2024 and does not provide real-time weather monitoring.

4. **No causal conclusions:** Differences in temperature and precipitation can be identified, but the analysis cannot establish that urbanization caused these differences.

5. **Urban heat island measurement:** The dataset contains daily weather observations, not direct measurements of urban and surrounding rural surface temperatures. Therefore, this project explores urban heat patterns rather than calculating formal Urban Heat Island intensity.

6. **No machine learning prediction:** This project focuses on data cleaning, exploratory data analysis, statistical summaries, and visualization. It does not build a machine learning prediction model.

Despite these limitations, the study provides useful insights into historical temperature and weather patterns across the selected Indian cities.


## Conclusion

This project analyzed historical daily weather data from ten major Indian cities between 2000 and 2024 using Data Science techniques. The dataset was inspected, checked for missing values and duplicates, and prepared for further analysis. Additional features such as year, month, season, weather condition, and daily temperature range were created to support the study.

Exploratory data analysis was used to examine city-wise temperature differences, yearly variations, monthly and seasonal patterns, precipitation, apparent temperature, and wind conditions. The results showed that weather patterns differed across cities and seasons. Summer recorded the highest average maximum temperature, while Winter recorded the lowest. Precipitation also showed clear seasonal variation, with higher average precipitation during the Monsoon season.

The visualizations and statistical summaries made it easier to identify patterns and compare environmental conditions across the selected cities. These findings support a better understanding of urban heat patterns based on historical weather observations.

The project demonstrates how Data Science can transform raw weather data into meaningful information through data preparation, exploratory analysis, and visualization. In the future, the study could be extended by including satellite-based surface temperature data, urban land-use information, and additional cities to investigate the Urban Heat Island effect more directly.
